## Part B: Cleaning

```python
print("Missing values count:")
print(df.isnull().sum())
```

```python
print("Duplicate rows count:")
print(df.duplicated().sum())
print("Removing Duplicates:")
print(df.drop_duplicates())
```

```python
print("column names:")
df.columns=df.columns.str.lower().str.replace(' ','_')
df.columns = df.columns.str.replace('-','_')
print(df.columns.tolist())
```

```python
print(f"changing 'order_date','ship_date' to datetime:")
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed')
df['ship_date'] = pd.to_datetime(df['ship_date'], format='mixed')
print(df['order_date'])
print(df['ship_date'])
```

```python
negative_sales = (df['sales'] <= 0).sum()
negative_quantity = (df['quantity'] <= 0).sum()

print("Sales <= 0:", negative_sales)
print("Quantity <= 0:", negative_quantity)
```

### Part B: Cleaning
Findings: 0 Missing Values, 0 Duplicates, Order Dates & Ship Dates converted to datetime, Column names converted to snake_case. No negative Sales or Quantity.


Part C:Feature engineering

```python


print("creating the new columns:")
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_day_name'] = df['order_date'].dt.day_name()

print(df['order_year'])
print(df['order_month'])
print(df['order_day_name'])
```

```python
print("Shipping Days")
df['shipping_days'] = (df['ship_date'] - df['order_date']).dt.days
print(df['shipping_days'])
```

```python
print("Profit Margin:")
df["profit_margin_%"] = ((df["profit"]/df["sales"])*100).round(2)
print(df["profit_margin_%"])
#check:
df['profit_margin_%'].describe()
```

```python
# is_loss(np.where)
df["is_loss"] = np.where(df["profit"] < 0, True,False)
print(df["is_loss"])
#check:
print(df["is_loss"].sum())
```

```python
#discount_category:
bins = [-0.01,0,0.2,0.5,1.0]
labels =['No discount','Low', 'Medium', 'High']
df['discount_category']= pd.cut(df['discount'], bins=bins, labels=labels)
print(df['discount_category'])
# Output check 
print(df[['discount', 'discount_category']].head(10))
print(f"Category Counts:")
print(df['discount_category'].value_counts())
```

### Part C Summary: Feature Engineering

New columns created:

- `order_year`, `order_month`, `order_day_name`: extracted from `order_date` to analyse trends over time.
- `shipping_days`: `ship_date - order_date`. Range is 0 to 7 days.
- `profit_margin_pct`: `(profit / sales) * 100`, rounded to 2 decimals. Negative for loss orders (minimum -275%).
- `is_loss`: 1 if `profit < 0`, else 0 (created with `np.where`). 1,871 loss-making rows.
- `discount_category`: discount binned with `pd.cut` into No discount (0), Low (up to 0.2), Medium (0.2 to 0.5), High (above 0.5).
