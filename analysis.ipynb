```python
import pandas as pd
import  numpy as np

df = pd.read_csv('Superstore.csv', encoding='latin-1')
df.head()
```

## Part B: Cleaning

```python
print("Missing values count:")
print(df.isnull().sum())
```

```python
print("Duplicate rows count:")
print(df.duplicated().sum())
print("Removing Duplicates:")
print(df.drop_duplicates())
```

```python
print("column names:")
df.columns=df.columns.str.lower().str.replace(' ','_')
df.columns = df.columns.str.replace('-','_')
print(df.columns.tolist())
```

```python
print(f"changing 'order_date','ship_date' to datetime:")
df['order_date'] = pd.to_datetime(df['order_date'], format='%d-%m-%Y')
df['ship_date'] = pd.to_datetime(df['ship_date'], format='%d-%m-%Y')
print(df['order_date'])
print(df['ship_date'])
```

```python
negative_sales = (df['sales'] <= 0).sum()
negative_quantity = (df['quantity'] <= 0).sum()

print("Sales <= 0:", negative_sales)
print("Quantity <= 0:", negative_quantity)
```

### Part B: Cleaning
Findings: 0 Missing Values, 0 Duplicates, Order Dates & Ship Dates converted to datetime, Column names converted to snake_case. No negative Sales or Quantity.


Part C:Feature engineering

```python


print("creating the new columns:")
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_day_name'] = df['order_date'].dt.day_name()

print(df['order_year'])
print(df['order_month'])
print(df['order_day_name'])
```

```python
print("Shipping Days")
df['shipping_days'] = (df['ship_date'] - df['order_date']).dt.days
print(df['shipping_days'])
```

```python
print("Profit Margin:")
df["profit_margin_%"] = ((df["profit"]/df["sales"])*100).round(2)
print(df["profit_margin_%"])
#check:
df['profit_margin_%'].describe()
```

```python
# is_loss(np.where)
df["is_loss"] = np.where(df["profit"] < 0, True,False)
print(df["is_loss"])
#check:
print(df["is_loss"].sum())
```

```python
#discount_category:
bins = [-0.01,0,0.2,0.5,1.0]
labels =['No discount','Low', 'Medium', 'High']
df['discount_category']= pd.cut(df['discount'], bins=bins, labels=labels)
print(df['discount_category'])
# Output check 
print(df[['discount', 'discount_category']].head(10))
print(f"Category Counts:")
print(df['discount_category'].value_counts())
```

### Part C Summary: Feature Engineering

New columns created:

- `order_year`, `order_month`, `order_day_name`: extracted from `order_date` to analyse trends over time.
- `shipping_days`: `ship_date - order_date`. Range is 0 to 7 days.
- `profit_margin_pct`: `(profit / sales) * 100`, rounded to 2 decimals. Negative for loss orders (minimum -275%).
- `is_loss`: 1 if `profit < 0`, else 0 (created with `np.where`). 1,871 loss-making rows.
- `discount_category`: discount binned with `pd.cut` into No discount (0), Low (up to 0.2), Medium (0.2 to 0.5), High (above 0.5).

***Part F: Insights and recommendations***

### Key Findings

* **Overall performance:** Total sales are 2.30M and total profit is 286K from 5,009 orders, giving an overall profit margin of 12.47%.
* **Yearly growth:** Sales grew from 484K (2011) to 734K (2014), with a small dip in 2012 (-2.83%). Profit grew every year, from 49.5K to 93.5K.
* **Seasonality:** Sales peak in November (349K), December (332K) and September (310K). February is the weakest month (60K).
* **Discounts hurt profit:** Orders with no discount earn +66.9 profit on average, but orders with more than 20% discount make a loss (-109.5 and -89.4). Discount and profit have a negative correlation (-0.22).
* **Loss-making areas:** Tables (-17.7K) and Bookcases (-3.5K) lose money. Central Furniture is the only loss-making region and category combination (-2,871). Texas (-25.7K), Ohio (-17.0K) and Pennsylvania (-15.6K) are the worst states.
* **Sales is not always profit:** Top-selling products like Cisco TelePresence EX90 (-1.8K) and GBC DocuBind P400 (-1.9K) lose money. Also, 18.72% of rows (1,871 of 9,994) are loss-making.

### Key Findings:

* **Peak Season:** Keep more stock and run marketing in September, November and December, because sales are highest in these months (310K, 349K, 332K). Do not add extra discounts, as demand is already high.
* **Central Furniture:** Furniture in the Central region is the only loss-making combination (-2,871). Its average discount is 29.7%, while other regions have 12% to 15%. Reduce discounts here, mainly on Tables, Bookcases and Furnishings.
* **Profit Margin:** Sales grew from 484K (2011) to 734K (2014), but the margin dropped from 13.43% to 12.74% in 2014. Track margin along with sales.
* **Discount Limit:** Keep discounts at 20% or below. Orders with no discount earn +66.9 profit on average, but orders above 20% discount lose money (-109.5 and -89.4).
